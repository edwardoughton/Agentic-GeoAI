## 1. What makes this a single-agent system?

The chat panel can already host an agent. The interface is not the defining feature.

| Approach | Who selects the next action? |
|---|---|
| Research conversation | You ask the next question |
| Human-directed analysis | You specify the sequence; AI helps implement it |
| Bounded single agent | One agent selects permitted actions in response to evidence |
| Multi-agent research | Multiple agent contexts exchange work through defined handoffs |

Our loop is **objective, search, inspect, record evidence, select next action, revise, stop**.

One local research agent handles search and synthesis. The later phone assistant reviews its exported artifacts; it is not an automated multi-agent coordination loop. The research agent is supervised and runs in an interactive session. This lab does not build an unattended Python/API agent.

## 5. Scaffold the agent's research state

Give the local coding agent this setup prompt. This creates files; it does not yet perform the investigation.

```text
In this repository, create the following structure without overwriting existing files:
README.md
AGENTS.md
research_brief.md
prompts/investigate.md
reports/latest.md
runs/
decisions/
.gitignore

README.md should explain the local-agent - GitHub - ChatGPT voice handoff.
Ignore .env files, local environments, caches, and credentials.
Leave research outputs explicitly marked as not yet investigated.
Do not create invented papers or results. Do not push yet.
I will supply the contents of the research rules, brief, and investigation prompt next.
```

Files define instructions and state; `AGENTS.md` does not instantiate an agent. Explicitly ask your agent to read it even if your tool normally discovers it automatically.

Paste the templates in the following sections into the corresponding files, using the editor or instructing the agent to save the exact text.

## 6. Research rules: `AGENTS.md`

```markdown
# Scientific research rules
- Use one research agent. Do not spawn subagents for this exercise.
- Never invent sources, citations, data, quotations, or findings.
- Separate evidence, interpretation, candidate gaps, and proposed studies.
- Record searches, tool actions, access failures, and concise reasons for changes.
- Give papers stable IDs: P01, P02, etc.
- Record authors, title, year, DOI/URL, publication status, access level,
  and the section/page supporting each extracted claim.
- Label access: full-text sections inspected, abstract only, or metadata only.
- Do not infer detailed methods from a title or abstract that does not state them.
- Search absence is not proof that nobody has done something.
- Seek competing prior work; do not manufacture a gap for a preferred project.
- Retain previous runs and original evidence. Do not commit credentials or
  copyrighted full texts without permission.
- Human review is required before claiming novelty, causality, or generalization.
- Final reports identify the run, date, evidence limits, and next scientific action.
```


## 7. Set the objective: `research_brief.md`

```markdown
# Research brief
## Objective
Investigate existing research and operational products for monitoring U.S.
inland waterway freight activity and disruption. Propose a feasible small study.

## Questions
What has been measured, using which observations and validation?
What do existing official products already provide?
Which candidate contribution survives competing prior work?
What evidence would weaken our preferred proposal?

## Scope
Prioritize U.S. inland waterways; label international/maritime comparators.
A PortWatch-inspired prototype is an option to assess, not a required conclusion.
Inspect existing USACE monitoring and nowcasting before proposing novelty.

## Budget and stop
At most 6 search queries, 8 source retrieval attempts, and 4 retained papers.
Record failed retrievals and stop when the budget is exhausted or evidence is
sufficient for a bounded next step. Ask for help if essential tools are unavailable.
Do not contact people, purchase data, or expand the scope without discussion.

## Outputs
Preserved evidence and activity log, candidate-gap assessment, and a spoken briefing.
No completed empirical results are required in this literature exercise.
```

For your own topic, replace the objective, questions and scope. Keep the budget, evidence requirements and stopping rules. Before consulting the agent, write your own candidate contribution and one piece of evidence that would undermine it.

## 8. Run one research agent: `prompts/investigate.md`

Save this prompt, then send it in your local agent conversation:

```text
Read AGENTS.md and research_brief.md. Conduct the bounded investigation using
available search, source-reading, and filesystem tools. Do not use subagents.
Inspect existing reports and decisions first if any exist.

Choose searches based on unresolved questions. Read sources, record evidence,
and change direction when findings challenge the initial idea. Keep within the
search/retrieval budget. If tools are unavailable, report that and stop rather
than inventing actions. Preserve a unique run directory under runs/.

Write these Markdown artifacts:
activity_log.md: actual queries/URLs, actions, access failures, short reasons for
search changes, budget counts, and stopping reason.
literature_evidence.md: paper IDs, verified metadata, access level, question,
geography, data, methods, findings, validation, and supporting section/page.
Mark unknown information rather than guessing.
gap_assessment.md: candidate studies, supporting AND competing prior work,
feasible data, unresolved assumptions, and what could invalidate the contribution.
briefing.md: a 500-650 word spoken overview, references and human decision questions.

Publish a self-contained copy of briefing.md at reports/latest.md. Include run ID,
date, a unique verification phrase, and paths to the preserved artifacts.
Do not claim verified novelty or completed experiments. Do not push yet.
```



## 9. Inspect the evidence before publishing

Open the output files. Check at least one source yourself.

| Check | What to inspect |
|---|---|
| Source exists | Open DOI/URL and match authors/title |
| Claim supported | Find the stated section or page |
| Access honest | Abstract-only extraction is identified |
| Gap bounded | Competing work and search limits included |
| Agent action observable | Log matches queries/tool calls you saw |
| Briefing accurate | Proposed work is not described as completed |

Require corrections if the agent claims a detailed method it never read. A bibliography is not evidence that the sources were inspected.


## 10. Commit and push the research record

In your research repository terminal, you can manually commit and push what you did:

```text
git status --short
git diff --check
git add README.md AGENTS.md research_brief.md prompts/investigate.md reports/latest.md runs .gitignore
git diff --cached --stat
git diff --cached
git commit -m "Publish first research investigation and voice briefing"
git branch -M main
git push -u origin main
```

Review staged content before committing. The commands assume the setup files and run artifacts exist. Stop on an error and diagnose it. Avoid staging unrelated files.

If Git asks for author identity, configure your chosen identity for this repository:

```text
git config user.name "YOUR NAME"
git config user.email "YOUR GITHUB EMAIL OR NOREPLY ADDRESS"
```

Then repeat the commit. If a push is rejected because the remote already has commits, ask your agent to inspect history and help reconcile it. Do not force-push as a shortcut.

Open your repository in the browser. Verify `reports/latest.md` displays your run ID and phrase.

## 12. Verify retrieval before starting Voice

Send this in a supported ChatGPT text conversation, replacing the repository name:

```text
Use the GitHub connection to read reports/latest.md in YOUR_USERNAME/research_agent.
State its run ID, date, and verification phrase, and cite the retrieved file.
Then read the matching literature_evidence.md and gap_assessment.md under runs/.
Tell me which files you actually retrieved. Do not infer their content if retrieval fails.
```

Compare the response with GitHub in your browser. Correct retrieval is established by matching the current report, not by a generic explanation of your topic.

If it cannot retrieve the file, check account, repository authorization, feature availability and file path. Some accounts support GitHub in research/agent experiences but not standard chat. Use the available supported experience or the instructor demonstration. Do not upload files as the default handoff for this lab.

## 13. Discuss the research verbally on your phone

Open the same AI conversation on iOS or Android. Select the **Voice icon**, allow microphone access and begin speaking. Use Live if it is available and supports the connected app for your account.

Start with:

> Give me a five-minute briefing based on the retrieved report. Distinguish verified findings, interpretations and candidate gaps. Explain paper IDs in plain language. Let me interrupt with questions. If the supplied evidence cannot answer something, say so.

Try these questions:

- Which paper most challenges my contribution?
- What did the agent actually read, and what is abstract-only?
- Explain P02's method only as far as the evidence allows.
- What evidence would make us abandon this proposal?
- Is this a scientific gap, a measurement problem, or an access problem?
- What should the first empirical experiment be?

Ask for source IDs/sections when questioning a claim. Spoken confidence does not replace checking the paper.

Voice/connected-app support varies. If direct GitHub access is unavailable in Voice, retrieve in text first and then discuss the retrieved context by voice. Enable **Settings -> Voice -> Background conversations** if you want to continue with the phone locked.


## 14. Return your decisions to the research agent

Ask your chosen phone assistant (ChatGPT or Claude):

> Summarize my proposed decisions, rejected suggestions, unresolved questions and next actions. Distinguish what I explicitly decided from your recommendations. Do not treat the voice transcript as a verbatim record.

Review and correct the summary. Save it locally as `decisions/<run-id>-human-review.md` with the report ID and date. You can copy this short decision summary; the research documents themselves remain in GitHub.

ChatGPT's documented GitHub app is read-only. The Claude import route in this lab supplies context rather than an automated decision-writing workflow. Voice discussion does not automatically edit files or launch your local research agent. This pilot has a deliberate human handoff.

Give the local agent:

```text
Read AGENTS.md, research_brief.md and decisions/<run-id>-human-review.md.
Explain the revised scope briefly, then execute the approved next investigation
within the existing budget. Preserve the prior run and create a new unique run.
Update reports/latest.md, changing the run ID and verification phrase.
Do not overwrite earlier evidence. Stop for unresolved scientific design decisions.
```

Review, commit and push the new run using explicit file paths. Ask ChatGPT to retrieve the new report and verify its ID; an earlier conversation may still contain an older version.

## 15. Troubleshooting the handoff

| Symptom | Next check |
|---|---|
| Agent invents a search | Compare the log with actual tool calls; remove unsupported claims |
| Source cannot be opened | Record failure/access level; narrow claims or choose another source |
| Git authentication fails | Resolve local Git sign-in; do not paste credentials into reports |
| Repo absent in ChatGPT | Check the connected GitHub account, allowed repo and organization approval |
| GitHub works in text, not Voice | Retrieve the evidence in text first, then continue by voice |
| Briefing describes the old run | Specify the path and ask for fresh retrieval plus run ID |
| Phone cannot see local changes | Save, commit, push, verify in GitHub, then retrieve again |
| Voice cannot answer a detailed question | Fetch the evidence register in a supported text session; do not fill gaps from memory |
| Feature unavailable | Complete research/publication and use a paired or instructor demo |

Consult the linked official documentation if settings have changed. Do not promise instant synchronization or automatic write-back.

For Claude: if a report is stale, refresh project GitHub content or re-add chat files and verify the run ID. If a new run is absent, update selected files. If Claude Code ignores shared rules, explicitly reference AGENTS.md and check CLAUDE.md.

## 1. Two connections with different purposes

| Connection | What you are doing | Where the files come from |
|---|---|---|
| GitHub integration | Reading published research artifacts | Pushed repository contents |
| Local Remote Control | Directing an agent that can act on local files | Connected computer/session |
| Cloud agent task | Running work in a cloud environment | Configured cloud checkout/environment |

For this lab, use local Remote Control. GitHub remains the publication and version-history layer. You do not need GitHub Actions.

Connecting the same repository does not necessarily transfer an existing conversation. Select the intended local project/session and verify it. Native agent management is different from viewing the whole desktop through remote-desktop software.



## 2. Route A: ChatGPT/Codex desktop pairing

Use the current ChatGPT desktop app on Windows or macOS and the mobile app, signed into the same account/workspace with Codex access.

1. On the host, open **Settings -> Connections -> Control this Mac or PC -> Set up/Add**.
2. Complete verification, scan the QR code with the phone, and finish pairing.
3. On mobile, open **Codex** or **Remote**, depending on the app label, and choose the host.
4. Select the local research project and an appropriate conversation; create one if necessary.

Availability varies by account/rollout; an administrator may need to enable access. Setup starts in the desktop app, not the CLI or IDE extension. If the setting is absent, check app/account/workspace and current documentation rather than changing experimental configuration flags.

Remote permits instructions, approvals and output review. Keep the host awake, online and its app running. Windows desktop Computer Use additionally needs an unlocked session; this file-based exercise does not require desktop automation.

[Official OpenAI Remote instructions](https://learn.chatgpt.com/docs/remote-connections)

## 4. Phone test 1: prove local reading

From your phone's connected agent conversation, send:

```text
This is a remote-access smoke test. Do not change files.
Report your actual working directory. Read AGENTS.md, research_brief.md,
and reports/latest.md if they exist. State the latest report's run ID and
one research constraint. If a file is missing, say so rather than inventing it.
```

Check the answer against your local editor. A generic description of your repository is not a pass. Confirm the directory and current file content.

For the instructor repository, the latest report may change; read the current run ID rather than hard-coding an expected answer into your test.

**Record:** provider, computer, project path, session name, time, and whether each file was available. Never record pairing URLs, credentials, or QR codes in the research repository.

## 5. Phone test 2: prove local writing and steering

Send this from the phone:

```text
Create remote_test/phone_check.md in this workspace, unless it already exists.
If it exists, preserve it and choose a new test filename. Include:
- verification phrase: phone to desk research test
- the working directory
- the run ID you actually read, or explicitly state unavailable
- status: connection test, not scientific evidence
Do not commit, push, delete files, or run research yet. Stop after writing.
```

Review any action approval. On the computer, open the newly created file. The file appearing locally is the evidence that execution reached the host.

Then send:

```text
Read the test file back. Suggest one useful next research action without executing it.
```

**Pass criteria:** correct local read; local artifact created; follow-up reaches the intended session. Pairing alone is insufficient.

Optional off-network test: repeat a read request over cellular data. If unavailable, record that result without claiming off-network access was verified.

## 6. Start and supervise a small scientific investigation

Use your existing brief or the inland-waterway example from Week 6A. From the connected phone session:

```text
Read and follow AGENTS.md and research_brief.md. Use one agent, no subagents.
Investigate one unresolved question from the latest report. Limit this run to
3 search queries and 4 source retrieval attempts. If browsing is unavailable,
report that rather than inventing a search.
Preserve a unique run directory. Write activity_log.md, literature_evidence.md,
gap_assessment.md and briefing.md; update reports/latest.md with a new run ID
and verification phrase. Separate evidence from interpretation. Do not claim
novelty or empirical results. Do not commit or push. Stop at the budget limit
or when enough evidence supports a bounded recommendation.
```

As work proceeds, inspect visible progress. At an appropriate point, send a steering message such as:

> Prioritize competing prior work. Do not expand to another corridor. Record this human intervention.

A remote status request can ask what has been inspected and what remains. Evaluate whether the log matches observable tool activity. A conversational account is not proof of a source inspection.

## 7. Approvals, stopping and publication

Review each requested action's purpose and scope. Remote access changes where you supervise; your normal permissions still matter. You do not need to broaden permissions to demonstrate agent behavior.

If scope drifts, use the app's available stop/cancel control. Confirm it stopped before starting conflicting work. Availability and labels differ; test the control with a small task if unsure.

Before publication, inspect source support, access labels and file changes. Use the local terminal or request these checks in the connected session:

```text
git status --short
git diff --check
git diff -- reports/latest.md
```

Untracked new run files will not appear in ordinary `git diff`: open them separately. Stage specific approved paths, review `git diff --cached`, then commit and push as taught in Week 6A. Tell the agent which publication actions you authorize.

Keep the connection-test artifact clearly marked or leave it uncommitted. A successful remote write says nothing about scientific accuracy.

## 8. Voice review and direct instructions

First confirm phone **text** management works. Voice availability in a particular remote task is a separate test; do not assume ordinary voice chat automatically controls the connected agent.

If a Voice control is offered in the connected conversation, ask for a briefing and verify the session/run before issuing an instruction. Otherwise use phone dictation to compose a prompt, review its transcript and send it. Dictation and two-way spoken conversation are different features.

For extended spoken research review, use the GitHub ? ChatGPT/Claude workflow in [Week 6A](06_01_single_research_agent_github_voice.ipynb). Return reviewed decisions to the connected local session. You now have two complementary routes: spoken evidence discussion and local agent management.

Example management message:

> Save my decision to decisions/<run-id>-human-review.md: focus on validating historical disruption alerts; defer vessel-to-tonnage estimation. Mark it as my approved scope. Propose the next experiment and wait for design review before execution.

Check the saved decision yourself. Only record what you actually approved; ask the assistant to distinguish its suggestions from your decisions.

## 9. Troubleshooting and recovery

| Problem | Next action |
|---|---|
| Phone paired, wrong files | Check selected host/project and actual working directory |
| Existing VS Code chat absent in Codex mobile | Open the intended project in the desktop route; do not assume conversation transfer |
| ChatGPT connection setting missing | Check current app, account/workspace and official setup; use a paired demonstration if unavailable |
| Claude command missing | Install via official setup; reopen terminal and verify version |
| Claude ineligible | Inspect login/subscription/admin policy and `claude doctor` |
| Phone shows progress but local artifact absent | Check execution location and whether the task was a cloud session |
| Host asleep/process closed | Restore host/app/process and reconnect; check task state before repeating |
| Approval pending | Review it on the supported screen; check before leaving a task unattended |
| Voice unavailable in remote session | Use reviewed dictation/text; use Week 6A for spoken artifact review |
| Report stale in GitHub discussion | Push approved changes and retrieve/refresh the new report ID |

Do not use undocumented feature flags or expose a public unauthenticated service to bypass a missing feature. Use supported pairing and account setup.

To end the lab, disconnect/revoke the phone through the host's connection controls or terminate the Claude Remote Control process when remote access is no longer needed. Verify the local session's state separately.

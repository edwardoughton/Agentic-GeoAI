## Git and GitHub are related, but different

Git is software for version control. This records changes to files as a history of snapshots called commits. Using this software we are able to inspect previous versions, work on branches, be more experimental, and then combine changes systematically.

GitHub is an online service that hosts Git repositories. It supports sharing, collaboration, issue tracking, code review, and backups of the shared project history.

A simple analogy:

| Concept | Role |
|---|---|
| Git | The version history system on your computer |
| GitHub | A website hosting a shared copy of that history |
| Repository | A project folder whose changes Git can track |
| Commit | A named snapshot in the project history |
| Branch | A parallel line of development |
| Clone | Create a local copy of an existing repository |
| Pull | Retrieve and integrate newer remote changes |
| Push | Send local commits to a remote repository |

### Why use Git for this class?

Agentic coding tools edit files quickly, sometimes across an entire project. Git gives us an inspectable history, making it easier to understand what changed and recover earlier work. It also supports reproducibility, so code, documentation, tests, and environment specifications can evolve together (in a single location we can readily access).

Git is not a substitute for judgment. Always inspect changes before committing them, especially changes produced by an AI agent.

## The command line

A shell is a program that reads text commands. A terminal is the window in which you interact with that shell. 

Here we will use Windows Command Prompt as the primary shell.

Every terminal has a current working directory, which is the folder in which a command operates unless another path is provided. 

Before running a command, check where you are.


## Essential navigation commands

| Purpose | Windows Command Prompt | macOS/Linux |
|---|---|---|
| Show current directory | `cd` | `pwd` |
| List files and folders | `dir` | `ls` |
| Enter a directory | `cd folder_name` | `cd folder_name` |
| Move up one directory | `cd ..` | `cd ..` |
| Return to your home directory | `cd /d %USERPROFILE%` | `cd ~` |
| Create a directory | `mkdir folder_name` | `mkdir folder_name` |
| Clear the terminal display | `cls` | `clear` |

Use quotation marks around a path containing spaces. The `/d` option also allows Command Prompt to change drive letters:

```bat
cd /d "C:\Users\YourName\Desktop\Course Work"
```

Press **Tab** while typing a path to autocomplete names and reduce spelling errors.

## Paths: relative and absolute

An absolute path describes a location from the filesystem root:

```text
C:\Users\YourName\Desktop\GeoAI
```

A relative path is interpreted from the current working directory:

```text
week2_datacenter_task\data
```

The symbol `.` means the current directory and `..` means its parent. Many command line errors occur because a command was run from the wrong directory.

## What is a Python virtual environment?

A virtual environment is an isolated Python installation for one project. It has its own interpreter and package collection.

We use one because:

* different projects may need different package versions
* installing packages globally can break unrelated projects
* a package requirements file makes the environment reproducible
* VS Code and Jupyter need to know which Python interpreter should run a notebook

The environment folder itself is specific to a machine and is not shared through GitHub. We share `requirements.txt`, then each student recreates the environment locally.

## Create the Week 2 environment

Make sure the terminal is in the repository root, so the location containing `requirements.txt`. 

Then run:

```bat
python -m venv .venv-agentic-geoai
```

On some Windows installations, the command may be `python3` instead of `python`. Check with:

```bat
python --version
```

The command creates `.venv-agentic-geoai`, which is ignored by Git because it should not be committed.

## Activate the environment

Windows Command Prompt:

```bat
.venv-agentic-geoai\Scripts\activate.bat
```

macOS/Linux:

```bash
source .venv-agentic-geoai/bin/activate
```

The terminal prompt should now begin with `(.venv-agentic-geoai)`. Activation changes which `python` and `pip` commands the terminal uses.


### If activation does not work

Confirm that Command Prompt is in the repository root and that the environment was created:

```bat
cd
dir .venv-agentic-geoai\Scripts
.venv-agentic-geoai\Scripts\activate.bat
```

If `activate.bat` is absent, recreate the environment with `python -m venv .venv-agentic-geoai`.


## Install the required packages

With the environment activated, run:

```bat
python -m pip install --upgrade pip
python -m pip install -r requirements.txt
```

Using `python -m pip` makes it explicit that packages are installed for the currently selected Python interpreter. The requirements file installs pandas, matplotlib, JupyterLab, and the Jupyter kernel support required this week.

## Verify the environment

```bat
python --version
python -m pip check
python -c "import pandas, matplotlib; print('Environment ready')"
```

`pip check` should report that no requirements are broken, followed by `Environment ready`.

You can inspect the active executable with:

```bat
python -c "import sys; print(sys.executable)"
```

Its path should include `.venv-agentic-geoai`.

## Useful package and environment commands

```bat
python -m pip list
python -m pip show pandas
python -m pip check
deactivate
```

These commands list installed packages, inspect pandas, check dependency consistency, and leave the virtual environment, respectively.

Do not install packages into whichever environment happens to be active without checking it first. Use `python -c "import sys; print(sys.executable)"` when uncertain.


## Command line safety

Before running a command suggested by an AI agent:

* Read the entire command.
* Check the current directory with `cd`.
* Understand which files or services it affects.
* Be especially cautious with commands that delete, overwrite, recursively move, install, or publish.
* Never paste passwords, API keys, or access tokens into notebooks or files tracked by Git.
* Ask for clarification when you cannot explain what a command will do.

Importantly, successful execution does not prove that a command was safe or appropriate, either by you or an agent.

## 🧭 1. A simple conceptual model of an agent

A conventional generative AI interaction is often approximately to prompt a model and then gain a response.

However, an agent instead works toward a goal through an iterative loop:

```text
                         ┌─────────────────────┐
                         │        GOAL         │
                         └──────────┬──────────┘
                                    ▼
                         ┌─────────────────────┐
                ┌───────►│    REASON / PLAN    │
                │        └──────────┬──────────┘
                │                   ▼
                │        ┌─────────────────────┐
   MEMORY /     │        │        ACT          │
   CONTEXT      │        │ tools / APIs / code │
                │        └──────────┬──────────┘
                │                   ▼
                │        ┌─────────────────────┐
                └────────│ OBSERVE / EVALUATE  │
                         └──────────┬──────────┘
                                    │
                              goal met?
                              /       \
                            no         yes
                            │           │
                            ▼           ▼
                          REVISE      REPORT
```

This is deliberately simple. We will use the same framework to compare agents in different domains.

### Research behind the model

Our diagram is a teaching synthesis, rather than a reproduction of one published architecture. 

The following research supports its main components:

| Part of our model | Research foundation | What it contributes |
|---|---|---|
| Reason / plan, act, observe | [Yao et al. (2023), ReAct](https://arxiv.org/abs/2210.03629), ICLR | Interleaves reasoning and actions, using observations from an environment to inform subsequent steps. |
| Memory, planning, and action | [Wang et al. (2024), agent survey](https://doi.org/10.1007/s11704-024-40231-1), Frontiers of Computer Science; [open preprint](https://arxiv.org/abs/2308.11432) | Organizes agent construction around profiling, memory, planning, and action, and reviews evaluation approaches. |
| Evaluate, revise, with memory | [Shinn et al. (2023), Reflexion](https://papers.nips.cc/paper_files/paper/2023/hash/1b44b878bb782e6954cd888628510e90-Abstract-Conference.html), NeurIPS | Uses feedback and stored verbal reflections to guide later attempts, without updating model weights. |
| Spatial workflow execution and verification | [Li and Ning (2023), Autonomous GIS](https://arxiv.org/abs/2305.06453), research paper/preprint | Proposes self-generating, self-organizing, self-verifying, self-executing, and self-growing GIS capabilities. |

These papers support different elements of the loop

They do not establish that every agent has all of these capabilities. For example, reflection can help revise an attempt, but does not by itself provide independent validation. 

Our explicit verification and human review requirements are course design choices for defensible geospatial analysis.

### 1.1 Core agent capabilities

| Capability | Question |
|---|---|
| Goal interpretation | What outcome is being requested? |
| Planning | What sequence of actions could achieve it? |
| Grounding / perception | What information exists in the environment? |
| Tool use | What software, APIs, databases, models, or instruments can be invoked? |
| Action | Can the system do something rather than only generate text? |
| Memory / state | What has already happened? |
| Evaluation | Did the action work? |
| Reflection / revision | Should the plan change? |
| Coordination | Should work be delegated to another agent/tool? |
| Verification | Is the result actually credible? |

**Key idea:** agents connect reasoning to action (e.g., we are giving an LLM access to specific functions to execute in the course of a prompt/set of prompts we provide).

Use these capabilities as questions to ask about a system. A system need not demonstrate every capability, and delegation alone does not establish greater autonomy.

For a broader taxonomy, read [Wang et al. (2024)](https://arxiv.org/abs/2308.11432). Our table adds explicit geospatial verification questions to the architectural components discussed in that survey.

### 1.2 We are dealing with a spectrum of autonomy rather than a binary distinction

```text
LOWER AUTONOMY                                           HIGHER AUTONOMY

Chatbot        Copilot       Tool-using       Workflow        Multi-agent /
                              agent             agent           autonomous system
   │              │              │                │                   │
answers        suggests       chooses          plans and           delegates,
questions      actions        actions/tools     iterates            reviews, adapts
```

For this course, an important transition is from:

**“Generate GIS code to calculate this metric...”**

to 

**“...here is a spatial problem, past papers/literature/repos, relevant data, and code/tools you may need to utilize. Help me work out how to investigate this problem, execute the analysis, evaluate the result, revise if necessary, and report back (including intermediate steps for verification/validation).”**

## 🔬 3. Agents in science

Scientific research is already iterative:

```text
OBSERVATION → QUESTION → HYPOTHESIS → EXPERIMENT / ANALYSIS
     ↑                                      │
     └──── INTERPRETATION ← RESULT ←────────┘
```

Agentic systems can potentially participate across several stages rather than performing only a single prediction.

Examples reported in the recent literature include:

- **biomedical discovery:** literature search, hypothesis generation, experiment planning, analysis, and hypothesis revision;
- **chemistry/materials:** literature retrieval, candidate selection, simulation, experimental planning, and interpretation;
- **experimental physics:** agents interacting with scientific instruments and adapting from observations;
- **scientific computing:** inspecting data, generating and running analyses, diagnosing failures, visualizing results, and comparing hypotheses.

> **Conceptual transition:** AI as a scientific model is now transitioning to AI as a participant in the scientific workflow.

See here: https://www.nsf.gov/funding/opportunities/ai-datasets-unlocking-dataset-value-ai-enabled-scientific-discovery 

The critical issue is not merely whether an agent can perform actions, but whether its resulting scientific claims are defensible and reproducible.

## 🗺️ 4. Why GIScience is a natural domain for agents

Geoinformation science (GIScience) brings together questions about spatial data, representation, analysis, and uncertainty. A store location problem connects these ideas in a practical workflow.

Consider the question:

> **Where should we locate a new home improvement store, such as a Home Depot?**

Note, this is a hypothetical classroom example, not a description of the company's actual site selection process.

Answering it may require the system to:

1. clarify the business objective and study area;
2. find household and housing data that could indicate potential demand;
3. obtain a road network and define customer travel time catchments;
4. locate competing home improvement stores and existing stores of the same chain;
5. choose appropriate coordinate systems and spatial units;
6. calculate accessibility for customers and delivery vehicles;
7. identify candidate parcels and inspect available land use constraints;
8. compare sites using explicitly defined demand, access, competition, and cost criteria;
9. map the alternatives;
10. check data coverage, calculations, and sensitivity to assumptions;
11. explain uncertainty and identify information still needed.

Household counts and housing characteristics are possible demand proxies, not observed sales. Good accessibility alone does not establish profitability or parcel availability. Students must define what ?best? means before delegating the analysis.

Traditional GIS can perform these operations. The agentic question is:

> **Can an agent decide which operations are required, orchestrate them, observe their results, verify them, and adapt its approach?**

This connection between spatial problem solving and workflow orchestration is central to [Li and Ning's Autonomous GIS framework (2023)](https://arxiv.org/abs/2305.06453).

## 🤖 5. Autonomous GIS and Agentic GeoAI

Li & Ning's [**Autonomous GIS**](https://arxiv.org/abs/2305.06453) framed an LLM as a reasoning core capable of orchestrating spatial data collection, analysis, visualization, and problem solving.

They proposed five goals:

| Goal | Meaning |
|---|---|
| Self-generating | Generate workflows/code |
| Self-organizing | Organize tasks, data, and operations |
| Self-verifying | Check intermediate and final results |
| Self-executing | Perform the workflow |
| Self-growing | Expand/improve capability |

```text
TRADITIONAL GIS                         AGENTIC GIS

Human defines workflow                  Human defines goal
        ↓                                       ↓
Human selects tools                     Agent plans workflow
        ↓                                       ↓
GIS executes                            Agent invokes GIS tools
        ↓                                       ↓
Human interprets                        Agent observes + revises
                                                ↓
                                         Agent verifies
                                                ↓
                                         Human reviews
```

Notice that self-verification is part of the concept from the outset.

## 🛰️ 7. Deep dive — Agentic Earth Observation

Remote sensing is naturally a multi-stage workflow:

```text
SCIENTIFIC QUESTION
        ↓
SENSOR / DATA SELECTION
        ↓
SEARCH + ACQUISITION
        ↓
QUALITY / CLOUD FILTERING
        ↓
PREPROCESSING
        ↓
MODEL / ALGORITHM SELECTION
        ↓
ANALYSIS
        ↓
SPATIOTEMPORAL COMPARISON
        ↓
VALIDATION
        ↓
INTERPRETATION
```

An EO agent can potentially **orchestrate this chain**, rather than simply classify one image.

### Example

> **Has there been significant forest loss around this protected area during the last three years?**

An agent might choose sensors and dates, retrieve imagery, handle clouds, create comparable observations, select a change-detection method, quantify change, validate against independent evidence, map the result, and report uncertainty.

The human supplies the **scientific goal**, rather than prescribing every remote-sensing operation.

## 🔎 8. Read a research example critically

[Li and Ning's Autonomous GIS](https://arxiv.org/abs/2305.06453) presents LLM-Geo, a prototype that uses an LLM to organize spatial data collection, analysis, and visualization. Its proposed capabilities are a research vision; a demonstration does not establish reliability for every GIS task.

[Earth-Agent](https://github.com/opendatalab/Earth-Agent) explores the orchestration of Earth-observation tools and specialist models. This gives us another way to think about an agent: it can organize existing analytical capabilities rather than replace them all with one model.

A remote-sensing model might classify an image. An EO agent may decide which observations to obtain, which tools to call, what comparisons to make, and when more evidence is needed.

When reading a paper, separate the proposed architecture, the tasks actually evaluated, and the capabilities that remain aspirations.

## ⚠️ 9. Reliability is still a major problem

Kao et al. (2026), [**Towards LLM Agents for Earth Observation**](https://aclanthology.org/2026.findings-acl.124/), evaluated agents on **408 Earth-observation questions** using the Google Earth Engine API.

Reported results included:

- **40.0% zero-shot accuracy**;
- generated code failing to execute in **more than 44%** of cases;
- reflexion-style approaches improving several strong-model results toward approximately **60%**.

The important lesson is:

> **Being able to call a geospatial tool is not equivalent to conducting reliable geospatial science.**

These figures describe the paper’s benchmark and experimental configurations, not every EO agent. Check what counts as a correct answer and distinguish execution failure from scientific error. This is one reason verification should not be treated as an optional final step.

## 💡 10. From research examples to commercial applications

Imagine a GIS platform that accepts: “Compare three possible warehouse locations.” It might retrieve layers, calculate accessibility, summarize alternatives, and generate a map.

Let us distinguish three possible behaviors:

| Behavior | What would you need to observe? |
|---|---|
| Assistant | It explains tools or suggests code for a human to execute. |
| Fixed workflow | It runs predefined steps using supplied parameters. |
| Agentic workflow | It chooses tools or revises steps in response to intermediate evidence. |

These are comparison criteria, not claims about a particular product. Product names and demonstrations alone do not establish which decisions a system can make or how reliably it makes them.

## 🎯 12. The key issue for this week: autonomy increases the verification burden

Compare:

**System A:** “Write Python code that calculates NDVI.”

**System B:** “Investigate whether drought conditions have worsened.”

System B may autonomously choose the:
- dataset;
- sensor;
- dates;
- study area;
- preprocessing;
- algorithm;
- threshold;
- validation strategy;
- interpretation.

Each autonomous decision creates another possible source of error.

```text
MORE AUTONOMY
     ↑
     │                        ● autonomous research
     │                  ● EO agent
     │            ● GIS agent
     │      ● coding copilot
     └────────────────────────────────────→
                   VERIFICATION BURDEN
```

## 📝 14. Take-home framework

For any agentic application, ask:

1. **GOAL** — What outcome is being pursued?
2. **ENVIRONMENT** — What can the agent observe?
3. **TOOLS** — What capabilities can it invoke?
4. **AGENCY** — Which decisions can it make autonomously?
5. **FEEDBACK** — How does it know whether an action worked?
6. **VERIFICATION** — How do we know whether the result is credible?
